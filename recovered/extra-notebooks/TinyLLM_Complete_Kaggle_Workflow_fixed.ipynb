# TinyLLM — Complete Kaggle Workflow (fixed)

This notebook automates the TinyLLM workflow: clone repo, install deps, train tokenizer, prepare dataset, and run small experiments.

```python
# 1) Clone or update repository
import os, subprocess
REPO_URL = 'https://github.com/ShaikKoduruAsif114/TinyLLM.git'
REPO_DIR = 'TinyLLM'
if os.path.exists(REPO_DIR):
    print('Repository exists — pulling latest')
    subprocess.run(['git','-C', REPO_DIR, 'pull'], check=False)
else:
    print('Cloning repository...')
    subprocess.run(['git','clone', REPO_URL], check=True)
print('Repo ready at', os.path.abspath(REPO_DIR))
```

```python
# 2) Quick inspection of key scripts
import os
root = 'TinyLLM'
for p in ['tokeniser/train_tokeniser.py','prepare_data.py','train.py']:
    print(p, 'exists=', os.path.exists(os.path.join(root,p)))
```

```python
# 3) Verify PyTorch and GPU availability
import torch
print('torch', torch.__version__, 'cuda_available:', torch.cuda.is_available())
if torch.cuda.is_available():
    try:
        print('GPU:', torch.cuda.get_device_name(0))
    except Exception as e:
        print('GPU query failed:', e)
```
